```
cd hctsa-master/hctsa
startup #navigate to the hctsa directory and load the packages for hctsa
```

```
TS_Init('ethoscope_metadata_1000.txt') #load the metadata file and preprocess
```

```
TS_Compute() # compute the files with hctsa
```

```
TS_Normalize('mixedSigmoid',[0.4,1.0], 'HCTSA.mat') #normalise data based on a mixed sigmoid model, with threshold of 0.4 for good time series and 1.0 for features 
```

```
TS_LabelGroups('HCTSA_1000.mat',{'ddt','lufenuron','flonicamid','permethrin','syn_cpd_1','abamectin','diamide_syn_cpd','chlorantraniliprole','cyantraniliprole','cyantraniliprole','dieldrin',
                'flubendiamide','imidacloprid','pymetrozine','ryanodine','spiroindoline','tetramethrin'}) #set group labels based on keywords
```

```
TS_Classify('HCTSA_1000.mat') #classify groups with a classification matrix
```
