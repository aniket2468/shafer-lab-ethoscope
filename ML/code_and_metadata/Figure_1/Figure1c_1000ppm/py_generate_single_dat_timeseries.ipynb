```python
#import your packages
import pandas as pd
import glob
import os
```

```python
#find your folder where your csv timeseries data are stored
path = r'/Documents/insecticide_data/your_timeseries_folder' # use your path
all_files = glob.glob(path + "/*.csv")
```

```python
#removes first two columns and header and gives a single timeseries of max velocity
for filename in all_files:
    df = pd.read_csv(filename, index_col=None, header=1)
    df.drop(df.columns[[0,1]], axis=1, inplace=True)
    df.to_csv(os.path.join(filename + ".dat"),index=False)
```
