```matlab
cd hctsa-master/hctsa
startup #navigate to the hctsa directory and load the packages for hctsa
```

```matlab
TS_Init('ethoscope_metadata.txt') #load the metadata file and preprocess
```

```matlab
TS_Compute() # compute the files with hctsa
```

```matlab
TS_Normalize('mixedSigmoid',[0.4,1.0], 'HCTSA.mat') #normalise data based on a mixed sigmoid model, with threshold of 0.4 for good time series and 1.0 for features 
```

```matlab
TS_Subset('HCTSA_N.mat',TS_GetIDs('1000','HCTSA_N.mat'),[],1,'HCTSA_1000.mat') #subset data to only files with 1000ppm keyword
```

```matlab
TS_Subset('HCTSA_N.mat',TS_GetIDs('100','HCTSA_N.mat'),[],1,'HCTSA_100.mat') #subset data to only files with 100ppm keyword
```

```matlab
TS_Subset('HCTSA_N.mat',TS_GetIDs('1','HCTSA_N.mat'),[],1,'HCTSA_1.mat') #subset data to only files with 1ppm keyword
```
